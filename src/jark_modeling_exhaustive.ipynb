# 1. Interpreting the Confusion Matrix
True Negatives (TN = 281)

False Positives (FP = 12)  
The model incorrectly flagged 12 good wavers as faulty.
→ extra inspection cost.

False Negatives (FN = 16)  
The model missed 16 faulty wavers.
→ this is dangerous and the most costly error.

True Positives (TP = 5)  
The model correctly identified 5 faulty wavers.

This shows the model struggles with detecting faulty wavers (class 1).

# 2. Explain Type 1 and Type 2 Errors
Type I Error (False Positive)  
Predict faulty when it is actually good.
→ cost = unnecessary inspection.

Type II Error (False Negative)  
Predict good when it is actually faulty.
→ cost = defective waver passes through the system.

In this problem, Type II errors are far more serious.

# 3. Explain why accuracy is misleading
Accuracy = 0.91 looks good, but:

93% of the test set is class -1

A dumb model predicting “always -1” gets 93% accuracy

Your model got 91%, which is actually worse than the baseline

Accuracy hides the model’s failure to detect faulty wavers.

Therefore, accuracy is not a meaningful metric for imbalanced datasets.

# 5. Write the theory sections (balancing, metrics, trade‑offs)
Oversampling
Duplicate minority class samples to balance the dataset.
Risk: overfitting.

Undersampling
Remove majority class samples.
Risk: losing information.

SMOTE
Synthetic Minority Oversampling Technique.
Creates synthetic minority samples between neighbors.
Reduces overfitting compared to simple oversampling.

ADASYN
Adaptive synthetic sampling.
Creates more synthetic samples in regions where the minority class is harder to learn.

ROSE
Random OverSampling Examples.
Generates synthetic samples using smoothed bootstrap.

## Performance Metrics
Accuracy  
(TP + TN) / total
Misleading for imbalanced data.

Sensitivity (Recall)  
TP / (TP + FN)
Ability to detect faulty wavers.
Most important metric here.

Specificity  
TN / (TN + FP)
Ability to detect good wavers.

Precision  
TP / (TP + FP)
Of all predicted faulty wavers, how many were correct.

F1‑score  
Harmonic mean of precision and recall.

ROC Curve  
Plots TPR vs FPR at different thresholds.

AUC  
Area under ROC curve.
Measures ranking quality.

## Why accuracy is not useful
Because the dataset is 93% class -1.
Accuracy is dominated by the majority class.

## Precision–Recall trade‑off
Increasing recall → more false positives → precision drops

Increasing precision → fewer false positives → recall drops

You cannot maximize both

You must choose based on cost

Here: recall is more important because missing a faulty waver is expensive
